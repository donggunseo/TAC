```python
import torch

ide = torch.load("./exp1/banking77/llama3.1-8b/42/indirect_effect.pt")
```

```python
k = 20

# 1) flatten해서 top-k 인덱스 구하기
values, flat_indices = torch.topk(ide.view(-1), k)

# 2) 2차원 (row, col) 인덱스로 변환
rows, cols = torch.unravel_index(flat_indices, ide.shape)

# 3) (row, col) 형태로 묶기
topk_indices_2d = list(zip(rows.tolist(), cols.tolist()))

print(topk_indices_2d)
```

```python
# (선택) 값까지 같이 보고 싶으면
topk_with_values = list(zip(topk_indices_2d, values.tolist()))
print(topk_with_values)
```
